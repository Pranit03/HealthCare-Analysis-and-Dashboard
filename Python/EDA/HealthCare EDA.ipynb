```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"D:\HealthCare Project\Data\Cleaned Data\Cleaned Data.healthcare.csv")
```

```python
df.head()
```

```python
blood_type =df['blood_type'].value_counts()
plt.figure(figsize=(7,7))

plt.pie(
    blood_type.values,
    labels=blood_type.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Blood Types by Patients")
plt.show()
```

```python
gender=df['gender'].value_counts()

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
        x="gender"
)

plt.title("Gender Wise Patients")
plt.xlabel("Genders")
plt.ylabel("No of Genders")
    
```

```python
billing_amount_inr=df['billing_amount_inr']

plt.hist(billing_amount_inr,bins=20)

plt.title("Billing Amount By Patient")
plt.xlabel("Bills")
plt.ylabel("Frequency")

plt.show()
```

```python

df["test_result"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Broad Result Distribution")
plt.xlabel("Test Status")
plt.ylabel("Number Test Result")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.boxplot(df['billing_amount_inr'])
plt.title("Billing Amount Distribution")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.histplot(df["blood_type"], bins=40, color="slateblue")
plt.title("Blood Types Distribution")
plt.xlabel("Blood Group")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```
