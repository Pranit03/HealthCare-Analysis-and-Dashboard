### Data Cleaning

```python
df['age'] =df['age'].median()
df['gender'] = df['gender'].fillna("Unknown")
df['blood_type'] =df['blood_type'].fillna("Unknown")
df['insurance_type']=df['insurance_type'].fillna("Unknown")
df['test_result']=df['test_result'].fillna("Unknown")
df['readmission_within_30_days']=df['readmission_within_30_days'].fillna("NA")
df['bmi']=df['bmi'].mean()
df['heart_rate_bpm']=df['heart_rate_bpm'].mean()
df['systolic_bp_mmhg']=df['systolic_bp_mmhg'].mean()
df['diastolic_bp_mmhg']=df['diastolic_bp_mmhg'].mean()
df['glucose_mg_dl']=df['glucose_mg_dl'].mean()
df['total_cholesterol_mg_dl']=df['total_cholesterol_mg_dl'].mean()
df['patient_satisfaction_score']=df['patient_satisfaction_score'].mean()
```
